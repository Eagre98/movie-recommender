# Movie Recommender System

A collaborative filtering recommendation engine built with Python and Pandas.
Input any movie title and get 5 similar movies based on rating patterns
from 610 real users across 9,742 films.

## How it works
Users who rated similar movies tend to have similar taste. This system
finds movies that were consistently rated alike by the same users —
no genre tags or movie descriptions needed, just math.

## Example output
recommend('Toy Story (1995)') → Toy Story 2, The Incredibles, Finding Nemo
recommend('Pulp Fiction (1994)') → Fight Club, Kill Bill, True Romance

## Dataset
MovieLens Small Dataset — 100,836 ratings, 9,742 movies, 610 users

```python
import pandas as pd
import numpy as np

print("Ready")
```

```python
# Load Files
movies = pd.read_csv('movies.csv')
ratings = pd.read_csv('ratings.csv')

print(f"Movies: {movies.shape[0]} films")
print(f"Ratings: {ratings.shape[0]} ratings")

movies.head()
```

```python
# Join the two tables together
data = ratings.merge(movies, on= 'movieId')
print(f"Combined table shape: {data.shape}")
data.head()
```

```python
# Build the movie matrix
movie_matrix = data.pivot_table(
    index='userId',
    columns='title',
    values='rating'
)

print(f"Matrix shape: {movie_matrix.shape}")
movie_matrix.head()
```

```python
#Calculate similarity bettween all movies
movie_similarity = movie_matrix.corr(method='pearson', min_periods=50)

print(f"Similarity matrix shape: {movie_similarity.shape}")
```

```python
def recommend(movie_title, n=5):
  if movie_title not in movie_similarity.columns:
    print(f"Movie not fount. Check the title perlling.")
    return

  similar = movie_similarity[movie_title].dropna()
  similar = similar.drop(movie_title)
  similar = similar.sort_values(ascending=False)

  print(f"\nBecause you liked '{movie_title}', you might enjoy:\n")
  for i, (title, score) in enumerate(similar.head(n).items(), 1):
    print(f" {i}. {title} (similarity: {score:.2f})")
```

```python
#test
recommend('Toy Story (1995)')
```
